## Imports

```python
import pandas as pd
from collections import defaultdict
```

## Load Dataset

```python
TARGET = 2500

samples = defaultdict(list)
counts = defaultdict(int)

with pd.read_csv('../datasets/news_cleaned_2018_02_13.csv',
    chunksize = 100_000,
    usecols = ["domain", "type", "title", "content"],
    engine = "python"
) as reader:
    for chunk in reader:
        for label in chunk["type"].dropna().unique():

                if counts[label] >= TARGET:
                    continue

                remainder = TARGET - counts[label]
                subset = chunk[chunk["type"] == label].head(remainder)
                samples[label].append(subset)
                counts[label] += len(subset)

        if all(c >= TARGET for c in counts.values()):
                break

    all_frames = []
    for label, frames in samples.items():
        all_frames.extend(frames)

    balanced_df = pd.concat(all_frames)

    print(balanced_df.shape)
    print(balanced_df["type"].value_counts())
    print(balanced_df.head())

    balanced_df.to_csv("../datasets/news_sample_balanced.csv", index=False)
        
```
