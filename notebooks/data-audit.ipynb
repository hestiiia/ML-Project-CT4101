Since the full dataset from [Fake News Corpus](https://github.com/several27/FakeNewsCorpus) contains over 9 million records with a severe type distribution disparity, I previously sampled the original dataset to collect 2500 records per type, leading to an overall dataset size of 30,000 records (see `balanced-sampler.ipynb` for more information).

## Removing redundant columns
Here, I'm dropping columns from the dataset I've judged to be redundant.

`keywords`, `summary` and `source`have been removed because they are solely comprised of NaN values. 

`Unnamed: 0` appears to be how Pandas renders an empty column contained in the dataset. I've removed this column.

`scraped_at`, `inserted_at`, `updated_at` is metadata with info on when the tuple was added or updated in the corpus, which is irrelevant to the task at hand.

From all I have observed, more than 75% (22,680) of the values of `meta_keywords` are made up of empty strings. For this reason, I've removed it from the dataset.

I've also dropped `id` which acts as an index to each article, because it's not necessary.

```python
len(data.columns)
```

```python
data["meta_keywords"].value_counts().head(10)
```

```python
columns_to_keep = ["domain", "type", "url", "title", "content", "authors", "tags", "meta_description"]

columns_to_drop = ["id", "keywords", "summary", "source", "Unnamed: 0", "scraped_at", "inserted_at", "updated_at", "meta_keywords"]
```

```python
print(f"Columns to keep: ", len(columns_to_keep))
print(f"Columns to drop: ", len(columns_to_drop))
```

```python
df = data[columns_to_keep]

df.shape
```

```python
df.tail(10)
```

## Inspecting columns that contain NaN values
Here I am checking the number of NaN values that remain in the dataset. 

As you can see, `authors`, `tags`, and `meta_description` contain a high volume of NaN values (>30%). For this reason, I've made the decision to drop these 3 columns, because they are too sparsely populated and imputation here would introduce a significant amount of synthetic data into the dataset.

`title` contains some NaN values but could be imputed in the preprocessing pipeline, so I've left it as is.

```python
nan = pd.DataFrame({
    "count": df.isna().sum(),
    "percent": (df.isna().mean() * 100).round(2)
})
nan.sort_values("percent")
```
